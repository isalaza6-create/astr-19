# Exponential Function Plot

This notebook creates a NumPy array ranging from 0 to 1, defines a function
that returns exp(x), and plots the result using matplotlib


```python
import numpy as np
import matplotlib.pyplot as plt
```

## Create x Array

This cell creates a NumPy array `x` with 100 evenly spaced values 
between 0 and 1

```python
x = np.linspace(0, 1, 100)
```

## Define exp(x) Function

This function returns the exponential of x using NumPy's exp function


```python
def exp_function(x):
    return np.exp(x)
```

## Compute y = exp(x) and Plot

This cell computes y using the defined function and plots x vs. y

```python
y = exp_function(x)

plt.plot(x, y)
plt.xlabel("Time [milliseconds]")
plt.ylabel("Awesomeness")
plt.title("Plot of exp(x)")
plt.show()
```
