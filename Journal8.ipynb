```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 1, 100)
```

```python
def sine(x):
    return np.sin(x)

def cosine(x):
    return np.cos(x)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))


axes[0].plot(x, sine(x), color='blue')
axes[0].set_title('sin(x)')
axes[0].set_xlabel('x')
axes[0].set_ylabel('sin(x)')


axes[1].plot(x, cosine(x), color='red')
axes[1].set_title('cos(x)')
axes[1].set_xlabel('x')
axes[1].set_ylabel('cos(x)')


plt.tight_layout()


plt.savefig('sine_cosine_plot.pdf')


plt.show()
```
