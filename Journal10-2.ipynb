# Random Number Distribution: Gamma Distribution

This notebook generates 1000 random numbers from the Gamma distribution 
(using NumPy), then creates a histogram with 100 bins and plots the result

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Generate Random Numbers

We generate 1000 random numbers from the Gamma distribution.

The Gamma distribution is controlled by:
- shape parameter (k)
- scale parameter (θ)

Here we choose:
- shape k = 2.0
- scale θ = 1.0

```python
shape = 2.0
scale = 1.0

random_numbers = np.random.gamma(shape, scale, 1000)
```

## Histogram of Gamma Distribution

This cell creates a histogram with 100 bins and plots the distribution.


```python
plt.hist(random_numbers, bins=100)
plt.xlabel("Random Value")
plt.ylabel("Frequency")
plt.title("Histogram of Gamma Distribution (1000 Samples)")
plt.show()
```
